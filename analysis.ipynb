```python
import csv

def parse_line_to_list(line):
    # Split the line into components based on spaces, handling cases with multiple spaces
    parts = line.split()
    # Process each part to correctly separate numeric values from text labels
    output = []
    buffer = []
    for part in parts:
        if part.replace('.', '', 1).isdigit():  # Check if the part is a number (allowing one dot for decimal)
            if buffer:
                output.append(' '.join(buffer))
                buffer = []
            output.append(part)
        else:
            buffer.append(part)
    if buffer:
        output.append(' '.join(buffer))
    return output

def convert_to_csv(data, output_filename):
    # Split the input data into lines
    lines = data.strip().split('\n')
    with open(output_filename, 'w', newline='') as csvfile:
        csv_writer = csv.writer(csvfile)
        # Write headers (optional, adjust as needed)
        csv_writer.writerow(['Subject Code', 'Subject Name', 'Number of Candidates', 'Grade Percentages'])
        # Process each line
        for line in lines:
            row = parse_line_to_list(line)
            csv_writer.writerow(row)

# Example data input
data = """
7127 ACCOUNTING ADV 2499 1.6 13.9 32.2 56.7 80.8 93.7 100.0
7205 ART & DESIGN ADV (3D DESIGN) 1025 11.0 23.4 52.3 77.8 91.9 97.7 100.0
7201 ART & DESIGN ADV (ART, CRAFT & DES) 3056 15.0 32.9 62.1 83.9 94.4 98.9 100.0
7202 ART & DESIGN ADV (FINE ART) 7283 16.8 36.5 67.4 88.8 96.8 99.4 100.0
7203 ART & DESIGN ADV (GRAPHIC COMM) 2911 14.2 33.0 65.7 87.7 96.2 99.3 100.0
7206 ART & DESIGN ADV (PHOTOGRAPHY) 5151 9.7 23.1 55.8 81.8 94.4 98.7 100.0
7204 ART & DESIGN ADV (TEXTILE DESIGN) 1972 14.4 35.2 69.0 89.8 97.4 99.5 100.0
7637 BENGALI ADV 17 41.2 52.9 76.5 88.2 94.1 94.1 100.0
7677 BIBLICAL HEBREW ADV 60 70.0 91.7 95.0 100.0 100.0 100.0 100.0
7402 BIOLOGY ADV 39274 8.6 26.8 48.4 69.5 85.6 95.1 100.0
7132 BUSINESS ADV 16355 3.8 15.2 43.6 72.7 90.8 97.7 100.0
7405 CHEMISTRY ADV 27662 9.0 31.3 52.1 70.7 85.3 95.2 100.0
7517 COMPUTER SCIENCE ADV 6280 7.0 27.1 45.5 65.1 83.8 96.0 100.0
7562 D&T: FASHION AND TEXTILES ADV 434 6.2 19.6 45.9 70.0 88.7 97.2 100.0
7552 D&T: PRODUCT DESIGN ADV 5114 5.9 18.2 45.1 70.6 88.2 96.6 100.0
7237 DANCE ADV 960 8.9 29.2 58.5 85.2 95.4 98.9 100.0
7262 DRAMA AND THEATRE ADV 2520 6.3 24.0 59.3 85.6 96.3 99.5 100.0
7136 ECONOMICS ADV 14825 7.1 28.9 55.4 78.4 91.8 97.9 100.0
7702 ENGLISH LANGUAGE ADV 9274 2.4 11.6 44.1 78.0 95.0 99.1 100.0
7707 ENGLISH LANGUAGE AND LITERATURE ADV 3236 4.3 13.7 43.8 76.9 94.6 98.9 100.0
7712 ENGLISH LITERATURE A ADV 7777 8.2 21.7 49.3 76.7 93.2 99.1 100.0
7717 ENGLISH LITERATURE B ADV 6879 8.1 22.3 49.2 77.0 94.0 99.1 100.0
7447 ENVIRONMENTAL SCIENCE ADV 1936 4.4 14.2 35.8 60.5 82.4 94.1 100.0
7652 FRENCH ADV 5282 12.1 38.4 68.6 87.2 95.5 98.7 100.0
7367 FURTHER MATHEMATICS ADV 1690 26.0 55.2 74.4 87.4 94.8 98.0 100.0
7037 GEOGRAPHY ADV 13961 5.8 23.6 53.2 77.8 92.6 98.2 100.0
7662 GERMAN ADV 1706 15.6 46.7 73.2 87.7 96.3 99.2 100.0
7042 HISTORY ADV 20338 5.7 24.2 52.6 79.0 93.9 98.8 100.0
7162 LAW ADV 5624 4.3 15.8 39.4 63.0 82.7 95.4 100.0
7357 MATHEMATICS ADV 14841 12.3 37.3 59.4 75.9 87.8 95.5 100.0
7572 MEDIA STUDIES ADV 2333 1.6 11.4 51.9 83.7 95.8 99.2 100.0
7672 MODERN HEBREW ADV 65 73.8 87.7 98.5 100.0 100.0 100.0 100.0
7272 MUSIC ADV 947 8.0 26.1 52.0 76.3 92.2 98.4 100.0
7682 PANJABI ADV 167 65.3 75.4 88.0 95.2 98.8 99.4 100.0
7172 PHILOSOPHY ADV 3819 4.7 21.6 48.6 72.8 87.2 96.3 100.0
7582 PHYSICAL EDUCATION ADV 5453 5.0 16.7 42.0 69.1 87.8 96.6 100.0
7408 PHYSICS ADV 22851 11.0 31.5 50.3 68.4 84.6 95.3 100.0
7687 POLISH ADV 927 13.1 57.8 81.2 92.6 97.8 99.1 100.0
7152 POLITICS ADV 4535 7.9 27.4 53.6 75.6 90.1 97.7 100.0
7182 PSYCHOLOGY ADV 64622 5.9 19.5 46.3 71.0 87.7 96.6 100.0
7062 RELIGIOUS STUDIES ADV 2871 4.9 22.7 50.5 74.0 90.0 97.6 100.0
7192 SOCIOLOGY ADV 37592 5.2 17.7 45.5 73.9 91.0 97.7 100.0
7692 SPANISH ADV 6071 11.3 37.2 65.1 84.4 95.3 98.8 100.0
"""

# Call the function with the data and the filename where the CSV will be saved
convert_to_csv(data, 'output.csv')
```
